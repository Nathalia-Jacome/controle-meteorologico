```python
import pandas as pd
import requests
import time

endpoint_planilha = "https://docs.google.com/spreadsheets/d/1VOome0iX2AnPVDtZucjqmHDqunatVe8w22--QDlOi7Q/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
def limpar_coordenada(valor):

    if pd.isna(valor):
        return None
    texto = str(valor).strip().replace(',', '.')

    if texto == '':
        return None

    if texto.count('.') > 1:
        parte_inteira, parte_decimal = texto.split('.', 1)
        texto = parte_inteira + '.' + parte_decimal.replace('.', '')

    try:
        return float(texto)
    except ValueError:
        return None
```

```python
print('===== Dados Coletados =====')
dados_planilha = pd.read_csv(endpoint_planilha)

#Correção das coordenadas com GPS - Padrão diferente
dados_planilha['Latitude:'] = dados_planilha['Latitude:'].apply(limpar_coordenada)
dados_planilha['Longitude:'] = dados_planilha['Longitude:'].apply(limpar_coordenada)
dados_planilha = dados_planilha.dropna(subset=['Latitude:', 'Longitude:'])

display(dados_planilha)
```

```python
# iterar cada resposta do formulário
for indice, linha in dados_planilha.iterrows():
  name_local = linha['Nome do local:']
  lat = linha['Latitude:']
  lon = linha['Longitude:']

  parametros = {
      'latitude': lat,
      'longitude': lon,
      'current':'temperature_2m'
  }

  resposta = requests.get(endpoint_api, params=parametros).json()

  display(resposta)
  print('\n\n')
```
